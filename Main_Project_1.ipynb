
# **AI & Social Media Impact on Student Health**


# **Problem Statement**

*   Students spend a large part of their day on social media and AI tools.
*   It is unclear how this usage affects their sleep, physical activity, and mental and physical health.
*   Schools and parents lack data-driven insight into which usage patterns are harmful.
*   This project analyzes student lifestyle data to measure these effects and predict health scores from daily habits.







# **Outcome**
The project helps uncover key patterns, trends, and hidden insights from the dataset. The analysis provides a clear understanding of the data and supports decision-making with evidence-based findings.

# **Domain**


*   **Education** (student well-being and digital behavior)




# Dataset Information

**Source:** Kaggle / Open Data Portal (https://www.kaggle.com/datasets/srisyra02/ai-and-social-media-impact-student-health-and-grades?resource=download)

**Year / Timeline:** Data collected 2025-2026

# Dataset Column/features Description:

**Student_ID** – Unique ID for each student

**Age** – Student age in years (13 to 25)

**Gender** – Male, Female, or Non-binary

**Education_Level** – High School, College, or University

**Daily_Social_Media_Hours** – Average hours on social media per day

**Daily_AI_Tool_Usage_Hours** – Average hours using AI tools per day

**Sleep_Hours** – Average hours of sleep per day

**Physical_Activity_Hours** – Average hours of physical activity per
day

**Mental-Health_Score**– Mental well-being score (higher is better)

**Physical_Health_Score** – Physical well-being score (higher is better)

**Correcting data types**

```
df["Gender"]= df["Gender"].astype("category")
```

 **Creating derived columns**

 **Total screen time and overall health**

```
Total_Screen_Time= df['Daily_Social_Media_Hours'] + df['Daily_AI_Tool_Usage_Hours']
Total_Screen_Time
```

```
df['Overall_Health_Score'] = (df['Mental_Health_Score'] + df['Physical_Health_Score']) / 2
```

**Adding a new colunm**

```
df['Age_Group'] = pd.cut(df['Age'], bins=[12, 15, 18, 21, 25],labels=['13-15', '16-18', '19-21', '22-25'])
```

```
df['Health_Category'] = pd.cut(df['Overall_Health_Score'], bins=[0, 60, 75, 100],
                               labels=['At Risk', 'Average', 'Healthy'])
```

```
df['Sleep_Category'] = pd.cut(df['Sleep_Hours'],bins=[0, 6, 8, 24],labels=['Short (<6h)', 'Normal (6-8h)', 'Long (>8h)'])
```

```
df['AI_Usage_Group'] = pd.cut(df['Daily_AI_Tool_Usage_Hours'], bins=[-0.1, 1, 3, 5, 24],
                              labels=['Low (0-1h)', 'Moderate (1-3h)', 'High (3-5h)', 'Very High (5h+)'])
```

# **Exploratory Data Analysis (EDA) And Visualization**

**Statistical** **summary**

```
df.describe()
```

```
col = "Daily_Social_Media_Hours"
x = df[col].dropna()

fig, ax = plt.subplots()
sns.histplot(data=df, x=col, bins=30, kde=True, ax=ax)
ax.axvline(x.mean(), linestyle="--", label="Mean = {x.mean():.2} h")
ax.set(title="Distribution of Daily Social-Media Use", xlabel="Hours per day", ylabel="Number of students")
ax.legend()
plt.tight_layout()
plt.show()
```

**1. Chart Explanation:**
The histogram visualizes the distribution of students' daily social media usage in hours. The bars represent the number of students in each usage interval, while the KDE curve illustrates the overall distribution pattern. The dashed vertical line indicates the mean daily usage.

**2. What the Chart Says:**
The chart helps us understand how social media usage varies among students, which usage intervals are most common, and how the mean and median describe their usage patterns.

**3. Features Used:**
The analysis also uses the mean, median, frequency of students, and KDE curve to understand the distribution.

**4. What We Learn:**
This visualization provides an overview of students' daily social media habits. It can help identify common usage ranges and inform digital-wellbeing initiatives. Further analysis is required to determine whether social media usage is associated with mental health or other student outcomes.

# **Final Story: AI & Social Media Impact on Student Health Dashboard**

Based on the dashboard created from our dataset of 16,000 students, here is the comprehensive data story summarizing how digital habits influence student well-being.

**1. Executive Summary & Key KPIs**
Our dashboard tracks the lifestyle habits and health outcomes of 16,000 students surveyed between 2025 and 2026. The high-level indicators reveal some standard benchmarks:

Total Sample Size: 16K students.

Average Social Media Usage: 4.54 hours per day.

Average Sleep: 6.55 hours per night (which falls slightly below
the recommended 7–8 hours for youth).

Average Overall Health Score: 80.26 out of 100 (a solid baseline, but with room for improvement).

2. Key Findings & Core Insights

A. AI Usage vs. Overall Health (The Tech-Utility Balance)

What the dashboard shows:

The chart Average of Overall_Health_Score by AI_Usage_Group reveals that health scores remain remarkably stable across different AI usage tiers. High-tech users (5h+ daily) maintain an average health score of 79, compared to 81 for low users (0-1h).

The Story:

Unlike social media, daily usage of AI tools does not show a sharp negative correlation with student health. This suggests that students may be utilizing AI productively (e.g., for education and task automation) without experiencing severe lifestyle disruption.

B. The Critical Role of Sleep in Overall Health
What the dashboard shows: In the Average of Overall_Health_Score by Sleep_Category chart, there is a clear, step-down trend:
Long Sleepers (>8h) boast the highest average health score of 85.
Normal Sleepers (6-8h) have an average health score of 81.
Short Sleepers (<6h) experience a noticeable drop to 77.
The Story: Sleep is the single most influential biological lever in this dataset. Restricting sleep to under 6 hours correlates directly with a decline in mental and physical health scores.

C. Consistent Social Media Habits Across Ages
What the dashboard shows: The Avg Social Media Hrs by Age_Group chart shows practically identical behavior across all age cohorts:
13-15 years: 4.5 hours/day
16-18 years: 4.5 hours/day
19-21 years: 4.5 hours/day
22-25 years: 4.6 hours/day

The Story:

Social media engagement is deeply normalized and completely uniform across a student's academic journey—from early high school through university graduation.

**D. Health Category Breakdown**

What the dashboard shows: The Total Students by Health_Category donut chart reveals:
76.25% (12K) of students are classified as Healthy.
22.87% (4K) are classified as Average.
A small portion is At Risk.

The Story:

While more than three-quarters of the student population is currently in a healthy state, nearly a quarter are sliding into "Average" or "At Risk" categories, requiring early intervention before lifestyle patterns solidify.

E. Demographic Composition

What the dashboard shows:

The academic distribution in the Count of Student_ID by Education_Level pie chart shows:
High School: 6.08K (38.02%)
University: 5.03K (31.42%)
College: 4.89K (30.56%)
The Story: The sample is well-balanced across education tiers, lending high credibility to the behavioral patterns observed.

**3. Strategic Recommendations**

Prioritize Sleep Campaigns: Schools and universities should focus on "sleep hygiene" rather than just screen-time limits, as sleep deprivation has the strongest correlation with lower overall health.
Healthy Social Media Guardrails: Since social media usage is consistently high (4.5+ hours/day) across all age groups, institutions should promote active, creative, or community-building digital habits rather than passive, endless scrollin.Encourage Responsible AI Productivity: Because AI tool usage does not severely harm health, educators should continue integrating AI tools into studies safely, showing students how to save time and reduce academic stress.
